# Sales Data


```python
sales_data = [
    {
        "date": "2026-01-05",
        "product": "Laptop",
        "category": "Electronics",
        "region": "North",
        "quantity": 2,
        "unit_price": 55000
    },
    {
        "date": "2026-01-12",
        "product": "Smartphone",
        "category": "Electronics",
        "region": "South",
        "quantity": 5,
        "unit_price": 25000
    },
    {
        "date": "2026-02-03",
        "product": "Tablet",
        "category": "Electronics",
        "region": "East",
        "quantity": 3,
        "unit_price": 18000
    },
    {
        "date": "2026-02-15",
        "product": "Headphones",
        "category": "Accessories",
        "region": "West",
        "quantity": 10,
        "unit_price": 2000
    },
    {
        "date": "2026-03-08",
        "product": "Laptop",
        "category": "Electronics",
        "region": "South",
        "quantity": 1,
        "unit_price": 55000
    },
    {
        "date": "2026-03-20",
        "product": "Smartwatch",
        "category": "Accessories",
        "region": "North",
        "quantity": 7,
        "unit_price": 5000
    },
    {
        "date": "2026-04-10",
        "product": "Monitor",
        "category": "Electronics",
        "region": "East",
        "quantity": 4,
        "unit_price": 12000
    },
    {
        "date": "2026-04-25",
        "product": "Keyboard",
        "category": "Accessories",
        "region": "West",
        "quantity": 8,
        "unit_price": 1500
    }
]

print("Sales data created successfully! ✅")
print(sales_data)
```

# Create Sales DataFrame

```python

import pandas as pd

df = pd.DataFrame(sales_data)

print("===== SALES DATAFRAME =====")
print(df)
```

# Calculate Sales Amount

```python
df["sales_amount"] = df["quantity"] * df["unit_price"]

print("Sales amount calculated successfully! ✅")

print(df[[
    "product",
    "quantity",
    "unit_price",
    "sales_amount"
]])
```

# Calculate Total Revenue

```python
total_revenue = df["sales_amount"].sum()

print("===== TOTAL SALES REVENUE =====")
print(f"Total Revenue: ₹{total_revenue:,.2f}")
```

# Product-wise Sales Analysis

```python
product_sales = (
    df.groupby("product")["sales_amount"]
    .sum()
    .sort_values(ascending=False)
)

print("===== PRODUCT-WISE SALES =====")
print(product_sales)
```

# Find Best-Selling Product

```python
best_product = product_sales.idxmax()
best_product_sales = product_sales.max()

print("===== BEST-SELLING PRODUCT =====")
print(f"Product: {best_product}")
print(f"Total Sales: ₹{best_product_sales:,.2f}")
```

# Category-wise Sales Analysis


```python
category_sales = (
    df.groupby("category")["sales_amount"]
    .sum()
    .sort_values(ascending=False)
)

print("===== CATEGORY-WISE SALES =====")
print(category_sales)
```


# Find Best-Performing Category



```python
best_category = category_sales.idxmax()
best_category_sales = category_sales.max()

print("===== BEST-PERFORMING CATEGORY =====")
print(f"Category: {best_category}")
print(f"Total Sales: ₹{best_category_sales:,.2f}")
```

# Region-wise Sales Analysis

```python
region_sales = (
    df.groupby("region")["sales_amount"]
    .sum()
    .sort_values(ascending=False)
)

print("===== REGION-WISE SALES =====")
print(region_sales)

```

# Find Best-Performing Region


```python
best_region = region_sales.idxmax()
best_region_sales = region_sales.max()

print("===== BEST-PERFORMING REGION =====")
print(f"Region: {best_region}")
print(f"Total Sales: ₹{best_region_sales:,.2f}")
```

# Monthly Sales Analysis

```python
df["date"] = pd.to_datetime(df["date"])

df["month"] = df["date"].dt.to_period("M").astype(str)

monthly_sales = (
    df.groupby("month")["sales_amount"]
    .sum()
)

print("===== MONTHLY SALES =====")
print(monthly_sales)
```

# Monthly Sales Trend Visualization

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Revenue (₹)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Product-wise Sales Visualization



```python
plt.figure(figsize=(8, 5))

plt.bar(
    product_sales.index,
    product_sales.values
)

plt.title("Product-wise Sales Revenue")
plt.xlabel("Product")
plt.ylabel("Sales Revenue (₹)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Category-wise Sales Pie Chart


```python
plt.figure(figsize=(8, 8))

plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Distribution by Category")

plt.show()
```

# Sales Amount Distribution Histogram


```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["sales_amount"],
    bins=5
)

plt.title("Distribution of Sales Amounts")
plt.xlabel("Sales Amount ($)")
plt.ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```

# Region-wise Sales Visualization

```python
plt.figure(figsize=(8, 5))

plt.bar(
    region_sales.index,
    region_sales.values
)

plt.title("Region-wise Sales Revenue")
plt.xlabel("Region")
plt.ylabel("Sales Revenue (₹)")

plt.tight_layout()
plt.show()
```

# Sales Performance Summary


```python
total_transactions = len(df)
total_quantity_sold = df["quantity"].sum()
average_transaction = df["sales_amount"].mean()

print("=" * 45)
print("       SALES PERFORMANCE SUMMARY")
print("=" * 45)

print(f"Total Transactions: {total_transactions}")
print(f"Total Quantity Sold: {total_quantity_sold}")
print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Average Transaction Value: ₹{average_transaction:,.2f}")
print(f"Best-Selling Product: {best_product}")
print(f"Best Category: {best_category}")
print(f"Best Region: {best_region}")

print("=" * 45)
```
