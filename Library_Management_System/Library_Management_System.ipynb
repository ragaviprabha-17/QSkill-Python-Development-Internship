```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
books = [
    {
        "book_id": 1,
        "title": "Python Programming",
        "author": "Mark Lutz",
        "category": "Programming",
        "available": True,
        "borrower": None
    },
    {
        "book_id": 2,
        "title": "Data Science Handbook",
        "author": "Jake VanderPlas",
        "category": "Data Science",
        "available": True,
        "borrower": None
    },
    {
        "book_id": 3,
        "title": "Artificial Intelligence",
        "author": "Stuart Russell",
        "category": "AI",
        "available": True,
        "borrower": None
    },
    {
        "book_id": 4,
        "title": "Machine Learning Basics",
        "author": "Andrew Ng",
        "category": "Machine Learning",
        "available": True,
        "borrower": None
    }
]

print("Book data created successfully! ✅")
```

```python
def view_books():

    print("\n===== ALL BOOKS =====")

    for book in books:

        print("\n" + "-" * 40)
        print(f"Book ID: {book['book_id']}")
        print(f"Title: {book['title']}")
        print(f"Author: {book['author']}")
        print(f"Category: {book['category']}")

        if book["available"]:
            print("Status: Available ✅")
        else:
            print("Status: Issued ❌")
            print(f"Borrower: {book['borrower']}")
```

```python
def add_book():

    book_id = len(books) + 1

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")
    category = input("Enter Book Category: ")

    new_book = {
        "book_id": book_id,
        "title": title,
        "author": author,
        "category": category,
        "available": True,
        "borrower": None
    }

    books.append(new_book)

    print("\n✅ Book added successfully!")
```

```python
def search_book():

    title = input("Enter book title to search: ")

    found = False

    for book in books:

        if title.lower() in book["title"].lower():

            print("\n" + "-" * 40)
            print(f"Book ID: {book['book_id']}")
            print(f"Title: {book['title']}")
            print(f"Author: {book['author']}")
            print(f"Category: {book['category']}")

            if book["available"]:
                print("Status: Available ✅")
            else:
                print("Status: Issued ❌")
                print(f"Borrower: {book['borrower']}")

            found = True

    if not found:
        print("\n❌ Book not found.")
```

```python
def issue_book():

    book_id = int(input("Enter Book ID: "))

    for book in books:

        if book["book_id"] == book_id:

            if book["available"]:

                borrower = input("Enter Borrower Name: ")

                book["available"] = False
                book["borrower"] = borrower

                print("\n✅ Book issued successfully!")

            else:

                print("\n❌ Book is already issued.")
                print(f"Borrower: {book['borrower']}")

            return

    print("\n❌ Book ID not found.")
```

```python
def return_book():

    book_id = int(input("Enter Book ID: "))

    for book in books:

        if book["book_id"] == book_id:

            if not book["available"]:

                book["available"] = True
                book["borrower"] = None

                print("\n✅ Book returned successfully!")

            else:
                print("\nBook is already available.")

            return

    print("\n❌ Book ID not found.")
```

```python
def library_statistics():

    total_books = len(books)

    available_books = sum(
        1 for book in books
        if book["available"]
    )

    issued_books = sum(
        1 for book in books
        if not book["available"]
    )

    print("\n===== LIBRARY STATISTICS =====")
    print(f"Total Books: {total_books}")
    print(f"Available Books: {available_books}")
    print(f"Issued Books: {issued_books}")
```

```python
def search_by_category():

    category = input("Enter category: ")

    found = False

    for book in books:

        if category.lower() in book["category"].lower():

            print("\n" + "-" * 40)
            print(f"Book ID: {book['book_id']}")
            print(f"Title: {book['title']}")
            print(f"Author: {book['author']}")
            print(f"Category: {book['category']}")

            if book["available"]:
                print("Status: Available ✅")
            else:
                print("Status: Issued ❌")
                print(f"Borrower: {book['borrower']}")

            found = True

    if not found:
        print("\nNo books found.")
```

```python
def search_by_author():

    author = input("Enter author name: ")

    found = False

    for book in books:

        if author.lower() in book["author"].lower():

            print("\n" + "-" * 40)
            print(f"Book ID: {book['book_id']}")
            print(f"Title: {book['title']}")
            print(f"Author: {book['author']}")
            print(f"Category: {book['category']}")

            if book["available"]:
                print("Status: Available ✅")
            else:
                print("Status: Issued ❌")
                print(f"Borrower: {book['borrower']}")

            found = True

    if not found:
        print("\nNo books found.")
```

```python
def save_books():

    books_df = pd.DataFrame(books)

    books_df.to_csv("library_books.csv", index=False)

    print("\nBooks saved successfully! ✅")
```

```python
def load_books():

    loaded_books = pd.read_csv("library_books.csv")

    print("\n===== LIBRARY BOOKS =====")

    print(loaded_books)
```

```python
while True:

    print("\n" + "=" * 50)
    print("      📚 LIBRARY MANAGEMENT SYSTEM 📚")
    print("=" * 50)

    print("1. View All Books")
    print("2. Add New Book")
    print("3. Search Book")
    print("4. Issue Book")
    print("5. Return Book")
    print("6. Library Statistics")
    print("7. Search by Category")
    print("8. Search by Author")
    print("9. Save Books to CSV")
    print("10. Load Books from CSV")
    print("11. Exit")

    choice = input("\nEnter your choice (1-11): ")

    if choice == "1":
        view_books()

    elif choice == "2":
        add_book()

    elif choice == "3":
        search_book()

    elif choice == "4":
        issue_book()

    elif choice == "5":
        return_book()

    elif choice == "6":
        library_statistics()

    elif choice == "7":
        search_by_category()

    elif choice == "8":
        search_by_author()

    elif choice == "9":
        save_books()

    elif choice == "10":
        load_books()

    elif choice == "11":
        print("\nThank you for using the Library Management System! 📚")
        break

    else:
        print("\n❌ Invalid choice! Please enter a number between 1 and 11.")
```
