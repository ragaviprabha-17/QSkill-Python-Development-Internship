# Student Data

```python
students = [
    {
        "name": "Ravi",
        "maths": 85,
        "science": 78,
        "english": 92
    },
    {
        "name": "Priya",
        "maths": 95,
        "science": 88,
        "english": 90
    },
    {
        "name": "Arun",
        "maths": 65,
        "science": 72,
        "english": 68
    },
    {
        "name": "Divya",
        "maths": 88,
        "science": 94,
        "english": 86
    },
    {
        "name": "Karthik",
        "maths": 55,
        "science": 60,
        "english": 58
    }
]

print("Student data created successfully! ✅")
print(students)
```

# Calculate Total and Average Marks


```python
for student in students:

    total = (
        student["maths"]
        + student["science"]
        + student["english"]
    )

    average = total / 3

    student["total"] = total
    student["average"] = average

print("Total and average marks calculated successfully! ✅")

for student in students:
    print(
        f"{student['name']} - "
        f"Total: {student['total']}, "
        f"Average: {student['average']:.2f}"
    )
```

# Assign Grades

```python
for student in students:

    average = student["average"]

    if average >= 90:
        grade = "A+"

    elif average >= 80:
        grade = "A"

    elif average >= 70:
        grade = "B"

    elif average >= 60:
        grade = "C"

    elif average >= 50:
        grade = "D"

    else:
        grade = "F"

    student["grade"] = grade

print("Grades assigned successfully! ✅")

for student in students:
    print(
        f"{student['name']} - "
        f"Average: {student['average']:.2f}, "
        f"Grade: {student['grade']}"
    )
```

# Pass / Fail Analysis


```python

PASS_MARK = 50

for student in students:

    if (
        student["maths"] >= PASS_MARK
        and student["science"] >= PASS_MARK
        and student["english"] >= PASS_MARK
    ):
        student["status"] = "Pass"
    else:
        student["status"] = "Fail"

print("Pass/Fail status calculated successfully! ✅")

for student in students:
    print(
        f"{student['name']} - "
        f"Grade: {student['grade']}, "
        f"Status: {student['status']}"
    )
```

# Find Top-Performing Student

```python
top_student = max(students, key=lambda student: student["average"])

print("===== TOP-PERFORMING STUDENT =====")

print(f"Name: {top_student['name']}")
print(f"Total Marks: {top_student['total']}")
print(f"Average: {top_student['average']:.2f}")
print(f"Grade: {top_student['grade']}")
print(f"Status: {top_student['status']}")
```

# Display Complete Student Performance


```python
print("===== STUDENT PERFORMANCE REPORT =====")

for student in students:

    print("\n" + "-" * 40)

    print(f"Name: {student['name']}")
    print(f"Maths: {student['maths']}")
    print(f"Science: {student['science']}")
    print(f"English: {student['english']}")
    print(f"Total Marks: {student['total']}")
    print(f"Average: {student['average']:.2f}")
    print(f"Grade: {student['grade']}")
    print(f"Status: {student['status']}")

print("\n" + "-" * 40)
```

# Create Student Performance DataFrame

```python
import pandas as pd

df = pd.DataFrame(students)

print("===== STUDENT PERFORMANCE DATAFRAME =====")
print(df)
```

# Subject-wise Average Analysis

```python
subjects = ["maths", "science", "english"]

subject_averages = {}

for subject in subjects:
    subject_averages[subject] = df[subject].mean()

print("===== SUBJECT-WISE AVERAGE =====")

for subject, average in subject_averages.items():
    print(f"{subject.capitalize()}: {average:.2f}")
```

# Subject-wise Average Visualization

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    subject_averages.keys(),
    subject_averages.values()
)

plt.title("Subject-wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

# Find Best-Performing Subject

```python
best_subject = max(
    subject_averages,
    key=subject_averages.get
)

print("===== BEST-PERFORMING SUBJECT =====")
print(f"Subject: {best_subject.capitalize()}")
print(f"Average Marks: {subject_averages[best_subject]:.2f}")
```

# Student Ranking


```python
ranking_df = df.sort_values(
    by="average",
    ascending=False
).reset_index(drop=True)

ranking_df.index = ranking_df.index + 1

print("===== STUDENT RANKING =====")

print(
    ranking_df[
        ["name", "total", "average", "grade", "status"]
    ]
)
```

# Final Class Performance Summary

```python
total_students = len(df)
passed_students = (df["status"] == "Pass").sum()
failed_students = (df["status"] == "Fail").sum()
class_average = df["average"].mean()

print("=" * 45)
print("       CLASS PERFORMANCE SUMMARY")
print("=" * 45)

print(f"Total Students: {total_students}")
print(f"Passed Students: {passed_students}")
print(f"Failed Students: {failed_students}")
print(f"Class Average: {class_average:.2f}")
print(f"Best Subject: {best_subject.capitalize()}")
print(f"Top Student: {top_student['name']}")

print("=" * 45)
```

# Class Performance Summary Chart

```python

summary_labels = [
    "Class Average",
    "Pass Percentage"
]

pass_percentage = (passed_students / total_students) * 100

summary_values = [
    class_average,
    pass_percentage
]

plt.figure(figsize=(8, 5))

plt.bar(
    summary_labels,
    summary_values
)

plt.title("Class Performance Summary")
plt.ylabel("Percentage / Average Marks")

plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```
